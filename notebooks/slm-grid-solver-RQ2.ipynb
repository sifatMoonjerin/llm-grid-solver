```python
from helper.results import load_json_file, write_json_file, create_result_summary
from helper.solver import solve_grids
from helper.prompts import generate_zero_shot_system_prompt
```

## SLM Models

```python
MODEL_QWEN_2B = "qwen3.5:2b"
```

## Load Data

```python
grids_5 = load_json_file("data/test_grids/grids_5.json")
grids_7 = load_json_file("data/test_grids/grids_7.json")
grids_10 = load_json_file("data/test_grids/grids_10.json")

grids_5 = grids_5[:2]
grids_7 = grids_7[:2]
grids_10 = grids_10[:2]
```

## Global Context with Self Correction

### 5x5 Grids

```python
result_details_global_self_correction_grid_5 = solve_grids(
    grids=grids_5, 
    model=MODEL_QWEN_2B,
    generate_system_prompt=generate_zero_shot_system_prompt,
    is_local_context=False, 
    is_self_correcting=True
)
```

```python
create_result_summary(result_details_global_self_correction_grid_5)
```

```python
write_json_file("results/RQ2/result_details_global_self_correction_grid_5.json", result_details_global_self_correction_grid_5)
```
