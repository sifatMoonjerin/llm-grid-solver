```python
import ollama
import json
```

```python
MISTRAL_MODEL = "mistral"
GEMMA_MODEL = "gemma:7b"
```

```python
SYSTEM_PROMPT = """
The robot is at position [2, 2] in a 5×5 grid.

The robot's position is represented as [row, column].

Movement rules:
- NORTH: [row - 1, column]
- SOUTH: [row + 1, column]
- EAST:  [row, column + 1]
- WEST:  [row, column - 1]

IMPORTANT:
- SOUTH changes ONLY the first number.
- EAST changes ONLY the second number.
- Never swap row and column.
- Apply exactly one movement.
- Return only the resulting position in the format [row, column].
"""
```

```python
def ask_slm(prompt, MODEL):
    print("Asking SLM...")
    response = ollama.chat(
        model=MODEL,
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": prompt}
        ],
        options={
            "temperature": 0.0, 
            "seed": 42
        }
    )

    return response["message"]["content"]
```

```python
print("Going North")
print(ask_slm("The robot executes the action North. What is its new position?", MISTRAL_MODEL))
print("Going South")
print(ask_slm("The robot executes the action South. What is its new position?", MISTRAL_MODEL))
print("Going East")
print(ask_slm("The robot executes the action East. What is its new position?", MISTRAL_MODEL))
print("Going West")
print(ask_slm("The robot executes the action West. What is its new position?", MISTRAL_MODEL))
```
