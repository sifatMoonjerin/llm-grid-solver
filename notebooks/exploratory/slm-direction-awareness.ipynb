```python
import ollama
import json
```

```python
MISTRAL_MODEL = "mistral"
GEMMA_MODEL = "gemma:7b"
```

```python
def ask_slm(system_prompt, user_prompt, MODEL):
    print("Asking SLM...")
    response = ollama.chat(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        options={
            "temperature": 0.0, 
            "seed": 42
        }
    )

    return response["message"]["content"]
```

```python
SYSTEM_PROMPT = """
The robot is at position [2, 2] in a 5×5 grid.

The robot's position is represented as [row, column].

Movement rules:
- NORTH: The robot moves up.
- SOUTH: The robot moves down.
- EAST:  The robot moves right.
- WEST:  The robot moves left.

Grid Indexing:
- The top-left corner of the grid is [0, 0].
- The bottom-right corner of the grid is [4, 4].

IMPORTANT:
- Apply exactly one movement.
- Return only the resulting position in the format [row, column].
"""
```

```python
user_prompt = "The robot executes the action North. What is its new position?"

print("Going North")
print(ask_slm(SYSTEM_PROMPT, user_prompt, GEMMA_MODEL))
```

```python
print("Going North")
print(ask_slm("The robot executes the action North. What is its new position?", GEMMA_MODEL))
print("Going South")
print(ask_slm("The robot executes the action South. What is its new position?", GEMMA_MODEL))
print("Going East")
print(ask_slm("The robot executes the action East. What is its new position?", GEMMA_MODEL))
print("Going West")
print(ask_slm("The robot executes the action West. What is its new position?", GEMMA_MODEL))
```
